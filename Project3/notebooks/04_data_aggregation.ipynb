```python
import pandas as pd

from parquet_files import COMPOSITE_DATA

composite_data = pd.read_parquet(COMPOSITE_DATA)

composite_data.head()
```

```python
avg_cols = ['custom_sentiment', 'finbert_sentiment']

static_cols = [
  'on_volatile_date',
  'adjacent_volatile_date',
  'very_near_volatile_date',
  'near_volatile_date',
  'p-7',
  'p-6',
  'p-5',
  'p-4',
  'p-3',
  'p-2',
  'p-1',
  'p+0',
  'p+1',
  'p+2',
  'p+3',
  'p+4',
  'p+5',
  'p+6',
  'p+7',
]

df_avg = (
    composite_data
    .groupby(['ticker', 'date'])
    .agg(
        {**{c: 'mean' for c in avg_cols},
         **{c: 'first' for c in static_cols}}
    )
    .reset_index()
)

print(len(df_avg))
(df_avg.head())
```

```python
from parquet_files import AGGREGATE_DATA


df_avg.to_parquet(AGGREGATE_DATA, engine='pyarrow', compression='gzip')
```
